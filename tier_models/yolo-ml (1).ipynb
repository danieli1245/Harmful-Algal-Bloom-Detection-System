```python
# YOLOv8 ML Pipeline for Algae Detection (High-Accuracy Version)
# Author: Kruthi (ML Component Only)

import os
from ultralytics import YOLO
import matplotlib.pyplot as plt

# ------------------------------------
# 1. Setup Dataset YAML
# ------------------------------------
yaml_path = "/kaggle/working/data.yaml"
yaml_content = """
path: /kaggle/input/algae-bloom-yolo-dataset/agle-bloom-1
train: train/images
val: valid/images

names:
  0: algae
  1: background
"""
with open(yaml_path, "w") as f:
    f.write(yaml_content)

# ------------------------------------
# 2. Load Model (YOLOv8-Large)
# ------------------------------------
model = YOLO("yolov8l.pt")  # or try yolov8x.pt for even more capacity

# ------------------------------------
# 3. Train Model with Optimized Hyperparameters
# ------------------------------------
model.train(
    data=yaml_path,
    epochs=150,
    imgsz=768,
    batch=8,
    name="algae_yolo_l_augmented",
    amp=True,
    auto_augment="randaugment",
    hsv_h=0.02,
    hsv_s=0.8,
    hsv_v=0.6,
    fliplr=0.5,
    flipud=0.2,
    translate=0.2,
    scale=0.6,
    copy_paste=0.1,
    cutmix=0.1,
    mixup=0.1,
    lr0=0.0005,
    lrf=0.005,
    momentum=0.95,
    weight_decay=0.0005,
    patience=30,
    val=True
)

# ------------------------------------
# 4. Evaluate Final Model (Optional)
# ------------------------------------
results = model.val()
results.plot()

# ------------------------------------
# 5. Predict on Custom Image
# ------------------------------------
image_path = "/kaggle/input/sample-algae/test1.jpg"  # Replace with actual test image
pred_results = model.predict(source=image_path, save=True, show_labels=True, show_conf=False)

# ------------------------------------
# 6. Show Prediction Output (Optional)
# ------------------------------------
from PIL import Image
output_dir = model.predictor.save_dir
output_image = os.path.join(output_dir, os.path.basename(image_path))
Image.open(output_image).show()
```
